```python
import glob
import pandas as pd

RAW_DIR = "/kaggle/input/datasets/shahiismyname/unsw-nb15-raw-dataset"
names = (pd.read_csv(f"{RAW_DIR}/NUSW-NB15_features.csv", encoding="latin-1")["Name"].str.replace(" ", "").str.strip().tolist())
for f in sorted(glob.glob(f"{RAW_DIR}/UNSW-NB15_RAW_FILES/*.csv")):
    d = pd.read_csv(f, header=None, names=names, dtype={"sport": str, "dsport": str}, low_memory=False)
    c = d["attack_cat"].fillna("Normal").str.strip().replace({"Backdoors": "Backdoor"})
    print(f.split("/")[-1], len(d), "cols", d.shape[1], "dups", int(d.duplicated().sum()))
    print((c.value_counts(normalize=True) * 100).round(3).to_dict())
```
